```python
# Question 7:
# Which is the most sold vehicle in this data?

print(data['Car_Name'].value_counts())
```

```python
# Question 11:
# How many single person owned vehicles are there?

print(data['Owner'].value_counts())
```

```python
# Question 12:
# Calculate depreciation
#Which is the most and least cost depreciated vehicle in data?
data['Depreciation'] = data['Present_Price'] - data['Selling_Price']

data.head()
```

```python
#Q 12
# Show vehicle depreciation

data[['Car_Name', 'Present_Price', 'Selling_Price', 'Depreciation']]
```

```python
#Q12
# Most depreciated vehicle

data.sort_values('Depreciation', ascending=False).head(1)
```

```python
# Least depreciated vehicle

data.sort_values('Depreciation').head(1)
```

```python
# Question 13:
# Which brands of vehicles are less affected by cost depreciation?

data[['Car_Name', 'Depreciation']].sort_values('Depreciation').head(10)
```

```python
#Question 14: Are there any factors that affect cost depreciation?
# Factor 1:Manufacturing Year
# Check whether manufacturing year affects depreciation

plt.scatter(data['Year'], data['Depreciation'])

plt.xlabel('Manufacturing Year')
plt.ylabel('Depreciation')
plt.title('Manufacturing Year vs Depreciation')

plt.show()
```

```python
# Check whether kilometers driven affects depreciation
#Factor2:Kilometers Driven
plt.scatter(data['Kms_Driven'], data['Depreciation'])

plt.xlabel('Kilometers Driven')
plt.ylabel('Depreciation')
plt.title('Kilometers Driven vs Depreciation')

plt.show()
```
